```python
file = "Mike-15-100m-Fly.txt"

def event_lookup(filename):
    conversions = {
    "Free": "freestyle",
    "Back": "backstroke",
    "Breast": "breaststroke",
    "Fly": "butterfly",
    "IM": "individual medley"
    }
    *_, distance, stroke = filename.removesuffix(".txt").split("-")
    lookup = f"{distance} {conversions[stroke]}"
    return lookup

print(event_lookup(file))
```
